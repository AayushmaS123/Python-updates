```python
# caesar_decrypt.py
# This program reads a file of encrypted log lines.
# It shifts each letter 3 places to the left to decrypt it.
# It saves every decrypted line to one file.
# It saves lines with the word BREACH to a second alert file.

SHIFT = 3

# Task 1
input_file = open("rawlogs.txt", "r")
master_file = open("decrypted_master.txt", "w")
alert_file = open("security_alerts.txt", "w")

# Task 2 and Task 3
for line in input_file:
    decrypted_line = ""

    for char in line:
        if char == "\n":
            decrypted_line = decrypted_line + char
        elif char == " ":
            decrypted_line = decrypted_line + char
        elif char >= "A" and char <= "Z":
            new_code = ord(char) - SHIFT
            if new_code < ord("A"):
                new_code = new_code + 26
            decrypted_line = decrypted_line + chr(new_code)
        elif char >= "a" and char <= "z":
            new_code = ord(char) - SHIFT
            if new_code < ord("a"):
                new_code = new_code + 26
            decrypted_line = decrypted_line + chr(new_code)
        else:
            decrypted_line = decrypted_line + char

    decrypted_line = decrypted_line.strip()
    master_file.write(decrypted_line + "\n")
    upper_line = decrypted_line.upper()
    if "BREACH" in upper_line:
        alert_file.write(decrypted_line + "\n")

# closing all files
input_file.close()
master_file.close()
alert_file.close()

print("Done. Check decrypted_master.txt and security_alerts.txt")
```
