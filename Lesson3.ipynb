```python
inventory = {
    "Wai Wai Noodles": 12,
    "Tokla Tea": 5,
    "Mustard Oil": 0,
    "Basmati Rice": 20,
    "Everest Masala": 8,
}

print("Starting inventory:", inventory)
```

```python
orders = [
    ["Wai Wai Noodles", 3],
    ["Mustard Oil", 2],
    ["Tokla Tea", 7],
    ["Basmati Rice", 15],
    ["Soap", 4],
    ["Everest Masala", 5],
    ["Wai Wai Noodles", 4],
    ["Basmati Rice", 8],
]

print("Number of orders:", (orders))
```

```python
print(inventory.get("Tokla Tea"))    
print(inventory.get("Mustard Oil"))  
print(inventory.get("Soap"))          
```

```python
fulfilled_orders = 0
unfulfilled_items = []
order_number = 0
```

```python
for order in orders:
    order_number += 1
    item = order[0]
    quantity = order[1]
    stock = inventory.get(item)

    if stock is None:
        print(f"ALERT: Order {order_number}: '{item}' is not in the inventory.")
        unfulfilled_items.append([item, quantity])
    elif stock >= quantity:
        inventory[item] = stock - quantity
        fulfilled_orders += 1
        print(f"SUCCESS: Order {order_number}: {quantity} x {item} fulfilled in full.")
    elif stock > 0:
        remainder = quantity - stock
        inventory[item] = 0
        unfulfilled_items.append([item, remainder])
        print(f"PARTIAL: Order {order_number}: only {stock} of {quantity} x {item} supplied. {remainder} short.")
    else:
        print(f"ALERT: Order {order_number}: '{item}' is out of stock.")
        unfulfilled_items.append([item, quantity])
```

```python
print("1. Final inventory:")
for item, count in inventory.items():
    print(f"   {item}: {count}")
print("   As a dictionary:", inventory)
```

```python
print("2. Fully fulfilled orders:", fulfilled_orders)
```

```python
print("3. Items not fully supplied:")
for entry in unfulfilled_items:
    print(f"   {entry[0]}: {entry[1]} short")
print("   As a list:", unfulfilled_items)
```
