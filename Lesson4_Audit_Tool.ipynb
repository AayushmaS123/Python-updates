```python
raw_amounts = [120.5, -30, 5, 0, 250, 9.99, 80, 10, 1000]
print("Raw amounts:", raw_amounts)
```

```python
def clean_transactions(amounts, min_threshold=10.0):
    return list(filter(lambda x: x > 0 and x >= min_threshold, amounts))
```

```python
cleaned = clean_transactions(raw_amounts)
print("Cleaned (default limit 10.0):", cleaned)

# we can also change the limit
print("Cleaned (limit 100):", clean_transactions(raw_amounts, 100))
```

```python
def apply_service_fee(amounts, fee_rate=0.02):
    return list(map(lambda x: round(x - x * fee_rate, 2), amounts))
```

```python
after_fee = apply_service_fee(cleaned)
print("After 2% fee:", after_fee)

# a different fee rate, 5%
print("After 5% fee:", apply_service_fee(cleaned, 0.05))
```

```python
final_list = apply_service_fee(clean_transactions(raw_amounts))
print("Final processed list:", final_list)
```

```python
def calculate_nested_total(data):
    # Base case 1: empty list
    if isinstance(data, list) and (data) == 0:
        return 0.0

    # Base case 2: a single number
    if isinstance(data, (int, float)):
        return data

    # Recursive step: non-empty list
    if isinstance(data, list):
        head = data[0]
        tail = data[1:]
        return calculate_nested_total(head) + calculate_nested_total(tail)

    # Anything else (like text) is not money, so it counts as zero
    return 0.0
```

```python
print("Empty list:", calculate_nested_total([]))
print("Single int:", calculate_nested_total(75))
print("Single float:", calculate_nested_total(49.5))
```

```python
nested = [100, [50, 25], [[10, 5], 40]]
print("Nested total:", calculate_nested_total(nested))
```

```python
deep = [1, [2, [3, [4, [5]]]]]
print("Deep list total:", calculate_nested_total(deep))

money = [1200.50, [350.25, [99.99]], [[10.5], 40]]
print("Float list total:", round(calculate_nested_total(money), 2))
```
